```python
import model_runner
import os
from io import BytesIO
from PIL import Image

# 테스트할 이미지 경로
test_image_path = os.path.join("test", "tacenol_01.jpg")

# 이미지 파일을 바이너리 모드로 읽기
with open(test_image_path, "rb") as f:
    image_bytes = f.read()

# ✅ BytesIO를 사용하여 이미지 변환 테스트
image = Image.open(BytesIO(image_bytes))
#image.show()  # 이미지가 정상적으로 표시되는지 확인

# YOLO 모델 실행 (알약 인식 테스트)
result = model_runner.detect_pill(image_bytes)
print(f"예측된 알약: {result}")
```
